1. Count how many men and women (feature sex) are represented in this dataset.

2. What is the average age of men (feature age) across the entire dataset?

**4-5. Calculate the mean and standard deviation of age for those who earn more than 50K per year (feature *salary*) and those who earn less than 50K per year.**

````python
less_than_50k = df[df['salary'] == '<=50K']
more_than_50k = df[df['salary'] == '>50K']
less_than_50k_age_std = less_than_50k['age'].std()
less_than_50k_age_mean = less_than_50k['age'].mean()
more_than_50k_age_std = more_than_50k['age'].std()
more_than_50k_age_mean = more_than_50k['age'].mean()
print(f"Less than 50k age mean: {less_than_50k_age_mean:.2f}")
print(f"Less than 50k age std: {less_than_50k_age_std:.2f}")
print(f"More than 50k age mean: {more_than_50k_age_mean:.2f}")
print(f"More than 50k age std: {more_than_50k_age_std:.2f}")
````

7. Display age statistics for each race (feature *race*) and gender. Use *groupby* and *describe*. In this way, find the maximum age of men of the *Asian-Pac-Islander* race.

8. Among whom is the proportion of high earners (>50K) greater among married or single men (feature *marital-status*)? Married individuals are considered those whose *marital-status* begins with *Married* (Married-civ-spouse, Married-spouse-absent, or Married-AF-spouse); the rest are considered single.

````python
high_earners = df[df['salary'].isin(['>50K']) &
                  df["sex"].isin(["Male"])].copy()
high_earners["marital-status"] = high_earners["marital-status"].apply(lambda x: "Married" if x.startswith("Married") else "Single")
high_earners = high_earners.groupby(by="marital-status")['age'].count()
high_earners = high_earners.rename('proportion')
high_earners_prop = high_earners / high_earners.sum()
high_earners_prop
````

9. What is the maximum number of hours a person works per week (feature *hours-per-week*)? How many people work that many hours, and what percentage of them earn a lot (>50K)?

````python
max_working_hours = df['hours-per-week'].max()
print(f"Max working hours: {max_working_hours}")
max_working_hours_people = df[df['hours-per-week'] == max_working_hours].groupby(by="salary").count()["age"]
max_working_hours_people_prop = max_working_hours_people / max_working_hours_people.sum()
max_working_hours_people_prop
````

10. Calculate the average working hours (*hours-per-week*) for low and high earners (*salary*) for each country (*native-country*).

````python
average_hours_per_country_salary = df.groupby(['native-country', 'salary'])['hours-per-week'].mean().unstack()
print(average_hours_per_country_salary)
````

11. Group people by age groups *young*, *adult*, *retiree*, where:
* *young* corresponds to 16-35 years
* *adult* - 35-70 years
* *retiree* - 70-100 years

Assign the name of the corresponding group for each person to a new column `AgeGroup`.

````python
ages = [16, 35, 70, 101]
labels = ['young', 'adult', 'retiree']
df['AgeGroup'] = pd.cut(df['age'], bins=ages, labels=labels, right=False)
df.head()
````

12-13. Determine the number of people earning >50K in each age group (AgeGroup column), and also output the name of the age group in which people more often earn more than 50K (>50K).

````python
age_group_salary_counts = df.groupby(['AgeGroup', 'salary'])['age'].count()
print(age_group_salary_counts.loc[(slice(None), '>50K')])
total_per_age_group = df.groupby('AgeGroup')['age'].count()
high_earners_per_age_group = age_group_salary_counts.loc[(slice(None), '>50K')]
proportion_high_earners = high_earners_per_age_group / total_per_age_group
print(proportion_high_earners)
age_group_most_high_earners = proportion_high_earners.dropna().idxmax()
print(age_group_most_high_earners)
````

14. Group people by employment type (occupation column) and determine the number of people in each group. Then write a filtering function `filter_func` that will return only those groups in which the average age (age column) is not more than 40 and in which all workers work more than 5 hours per week (hours-per-week column).

````python
employment = df.groupby(by="occupation")['age'].count()
def filter_func(employment: pd.DataFrame) -> pd.DataFrame:
  avg_age_by_employment = (df.groupby(by="occupation")["age"].mean() < 40)
  avg_hours_by_employment = (df.groupby(by="occupation")["hours-per-week"].mean() > 5)
  return employment[avg_age_by_employment & avg_hours_by_employment]
print(filter_func(employment))
````
